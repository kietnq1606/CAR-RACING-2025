```python
import os
import random
import shutil

# Define the ratio for splitting the data
train_ratio = 0.7
v
test_ratio = 0.1

# Create the train, valid, and test directories within the root
train_dir = os.path.join(root, 'train')
valid_dir = os.path.join(root, 'valid')
test_dir = os.path.join(root, 'test')

os.makedirs(train_dir, exist_ok=True)
os.makedirs(valid_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# Assuming your data is in a single directory structure within 'root'
# You might need to adjust this based on your actual data structure
data_dir = root # Or specify the actual directory containing your images and labels

# Get a list of all image files (assuming .jpg extension, adjust if needed)
image_files = [f for f in os.listdir(os.path.join(data_dir, 'images')) if f.endswith('.jpg')] # Look for images in the 'images' subdirectory
random.shuffle(image_files)

# Calculate the number of files for each split
num_images = len(image_files)
num_train = int(num_images * train_ratio)
num_val = int(num_images * val_ratio)
num_test = num_images - num_train - num_val # Ensure all files are included

# Split the image files
train_files = image_files[:num_train]
val_files = image_files[num_train:num_train + num_val]
test_files = image_files[num_train + num_val:]

# Function to copy files and their corresponding labels
def copy_files(file_list, dest_dir):
    os.makedirs(os.path.join(dest_dir, 'images'), exist_ok=True)
    os.makedirs(os.path.join(dest_dir, 'labels'), exist_ok=True)
    for fname in file_list:
        # Copy image
        src_img_path = os.path.join(data_dir, 'images', fname) # Correct source path for images
        dest_img_path = os.path.join(dest_dir, 'images', fname)
        shutil.copy2(src_img_path, dest_img_path)

        # Copy label (assuming label file has the same name with .txt extension)
        label_fname = os.path.splitext(fname)[0] + '.txt'
        src_label_path = os.path.join(data_dir, 'labelTXT', label_fname) # Correct source path for labels (assuming 'labelTXT')
        dest_label_path = os.path.join(dest_dir, 'labels', label_fname)
        if os.path.exists(src_label_path):
            shutil.copy2(src_label_path, dest_label_path)
        else:
            print(f"Warning: Label file not found for {fname}")


# Copy files to their respective directories
print("Splitting and copying files...")
copy_files(train_files, train_dir)
copy_files(val_files, valid_dir)
copy_files(test_files, test_dir)

print("Data splitting complete.")
print(f"Train images: {len(train_files)}")
print(f"Validation images: {len(val_files)}")
print(f"Test images: {len(test_files)}")
```

```python
import os, subprocess, sys, shutil

repo_dir = "/content/yolov5"
if not os.path.isdir(repo_dir):
    subprocess.run(["git", "clone", "https://github.com/ultralytics/yolov5", repo_dir], check=True)
    subprocess.run([sys.executable, "-m", "pip", "install", "-r", f"{repo_dir}/requirements.txt", "-q"], check=True)

os.environ["WANDB_MODE"] = "disabled"

data_yaml_path = os.path.join(root, "dataset.yaml")
weights_path = os.path.join(repo_dir, "yolov5n.pt") # Corrected weights path

# Kiểm tra tồn tại
assert os.path.isfile(data_yaml_path), f"Không thấy YAML: {data_yaml_path}"
assert os.path.isfile(weights_path), f"Không thấy weights: {weights_path}"

# 4) Lệnh train chuẩn YOLOv5
cmd = [
    sys.executable, "train.py",
    "--data", data_yaml_path,
    "--weights", weights_path,
    "--img", "320",
    "--epochs", "100",
    "--batch", "16",
    "--rect",
    "--device", "0",       # GPU
    "--project", "runs/train",
    "--name", "exp_signs"
]
print("Running:\n", " ".join(cmd))

# Chạy train từ thư mục repo, in stdout/stderr
res = subprocess.run(cmd, cwd=repo_dir, text=True, capture_output=True)
print("STDOUT:\n", res.stdout)
print("STDERR:\n", res.stderr)
if res.returncode != 0:
    raise SystemExit(f"Train failed with code {res.returncode}")
```

```python
import os, glob, shutil, datetime

RUNS = "/content/yolov5/runs/train"
OUT_DIR = "/content/drive/MyDrive/Car racing  2025/weight"  # nơi cần lưu

# Lấy exp mới nhất
exps = sorted(glob.glob(f"{RUNS}/*"), key=os.path.getmtime)
assert exps, f"Chưa thấy thư mục runs/train trong: {RUNS}"
exp = exps[-1]
wdir = os.path.join(exp, "weights")

best = os.path.join(wdir, "best.pt")
last = os.path.join(wdir, "last.pt")
assert os.path.isfile(best), f"Không tìm thấy best.pt trong {wdir}"
assert os.path.isfile(last), f"Không tìm thấy last.pt trong {wdir}"

stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M")
name  = os.path.basename(exp)  # ví dụ: exp_signs

BEST_OUT = os.path.join(OUT_DIR, f"{name}_{stamp}_best.pt")
LAST_OUT = os.path.join(OUT_DIR, f"{name}_{stamp}_last.pt")

shutil.copy2(best, BEST_OUT)
shutil.copy2(last, LAST_OUT)

print("✅ Saved:")
print(" -", BEST_OUT)
print(" -", LAST_OUT)
```
